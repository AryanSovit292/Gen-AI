```python
import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer
```

```python
model_name = "gpt2"

tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)

tokenizer.pad_token = tokenizer.eos_token

print("GPT-2 model loaded successfully!")
```

```python
def generate_text(prompt, **kwargs):

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    )

    output = model.generate(
        inputs["input_ids"],
        attention_mask=inputs["attention_mask"],
        max_new_tokens=80,
        pad_token_id=tokenizer.eos_token_id,
        **kwargs
    )

    generated_text = tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )

    return generated_text
```

```python
prompt = "The future of artificial intelligence will change the way humans live."
```

```python
output1 = generate_text(
    prompt,
    do_sample=True,
    temperature=1.0
)

print(output1)
```

```python
output2 = generate_text(
    prompt,
    do_sample=True,
    temperature=0.5
)

print(output2)
```

```python
output3 = generate_text(
    prompt,
    do_sample=True,
    temperature=1.5
)

print(output3)
```

```python
output4 = generate_text(
    prompt,
    do_sample=True,
    temperature=1.0,
    top_k=20
)

print(output4)
```

```python
custom_output3 = generate_text(
    custom_prompt,
    do_sample=True,
    temperature=1.3,
    top_p=0.95
)

print(custom_output3)
```

```python
# OBSERVATIONS:

1. Temperature (0.3):
   A low temperature kept the generated text focused and predictable. The model selected safer and more likely words, resulting in less creative and less random output.

2. Top-k (10):
   Limiting the model to the top 10 most probable word choices made the generated text more controlled and predictable. This reduced randomness in the output.

3. Top-p (0.5):
   Using a top-p value of 0.5 restricted the model to highly probable words. The generated text was clearer, more focused, and contained less randomness.

4. Repetition Penalty (1.5):
   The repetition penalty reduced the chances of the model repeating the same words or phrases. This made the generated text more diverse and natural.

5. Max New Tokens (50):
   This parameter controlled the maximum length of the generated output. It produced a shorter response without significantly changing the creativity or style of the text.
```
